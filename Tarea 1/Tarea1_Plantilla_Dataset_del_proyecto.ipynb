### 5.2 Duplicados y columnas vacías (5 puntos)

*Elimine las filas duplicadas exactas y las columnas totalmente vacías. Si su tabla no tiene identificador de fila, decida con el diccionario de datos si dos filas iguales pueden ser reales.*

```python
limpio = datos.drop_duplicates().reset_index(drop=True)
vacias = limpio.columns[limpio.isna().all()].tolist()
limpio = limpio.drop(columns=vacias)

print("Filas eliminadas:", len(datos) - len(limpio), "| columnas eliminadas:", vacias)
limpio.shape
```

*Decisión:* …

### 5.3 Texto: espacios, mayúsculas y sinónimos (5 puntos)

*Una celda por columna de texto con variantes. El orden importa: espacios → caja → sinónimos (`replace`). Al final, cada columna debe tener una sola forma de escribir cada valor.*

```python
limpio["ciudad"] = (limpio["ciudad"].str.strip().str.title()
                    .replace({"Gye": "Guayaquil", "Uio": "Quito", "Cue": "Cuenca", "Mec": "Manta", "Mch": "Machala"}))
limpio["ciudad"].value_counts(dropna=False)
```

```python
# ← adapte: una columna por celda, con sus propias variantes
limpio["categoria"] = (limpio["categoria"].str.strip().str.lower()
                       .replace({"electronica": "electrónica", "papeleria": "papelería"}).str.capitalize())
limpio["producto"] = limpio["producto"].str.strip().str.capitalize()
limpio["metodo_pago"] = limpio["metodo_pago"].str.strip().str.lower().replace({"transf.": "transferencia"})
limpio[["ciudad", "categoria", "producto", "metodo_pago"]].nunique()
```

*Decisión (qué variantes se unificaron y por qué):* …

### 5.4 Tipos: números y fechas (5 puntos)

*Convierta a número las columnas numéricas leídas como texto (`str.replace` para quitar símbolos, `pd.to_numeric(errors="coerce")`) y a fecha las columnas de fecha (`pd.to_datetime` con `format=`; un `to_datetime` por formato si hay varios). Compruebe después que `dtypes` muestra lo que dice el diccionario de datos.*

```python
precio = limpio["precio_unitario"].str.replace("$", "", regex=False).str.replace(",", ".", regex=False)
limpio["precio_unitario"] = pd.to_numeric(precio, errors="coerce")

iso = pd.to_datetime(limpio["fecha"], format="%Y-%m-%d", errors="coerce")
dmy = pd.to_datetime(limpio["fecha"], format="%d/%m/%Y", errors="coerce")
limpio["fecha"] = iso.fillna(dmy)

print("Fechas no convertidas:", limpio["fecha"].isna().sum())
limpio.dtypes
```
